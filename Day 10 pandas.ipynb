```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Store' : ['A','B','A','C','B','A'],
    'Product' : ['Laptop','Mouse','Tablet','Laptop','Mouse','Tablet'],
    'Sales' : [1000, 50, 300, 1200, 60, 350],
    'Quantity' : [5, 10, 3, 6, 8, 4]
})
print(df)
```

```python
print(df["Sales"].sum())
print(df["Sales"].mean())
print(df["Sales"].count())
print(df["Sales"].median())
print(df["Sales"].std())
```

```python
df["Sales"].agg(["sum","std","mean","count","median"])
```

```python
pd.DataFrame(df.groupby("Store"))
```

```python
df.groupby("Store")["Sales"].sum()
```

```python
print(df.groupby("Store")["Sales"].sum())
print(df.groupby("Store")["Sales"].mean())
print(df.groupby("Store")["Sales"].std())
print(df.groupby("Store")["Sales"].count())
```

```python
df.groupby("Store")["Sales"].agg(["sum","std","mean","count","median"])
```

```python
df.groupby("Store")["Quantity"].agg(["sum","std","mean","count","median"])
```
