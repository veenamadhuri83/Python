# Decorators

```python
# Def : A Decorator is a function that takes another function as input and
#       add the EXTRA functionality without changing the original function code.


```

```python
# Example-1:
def decorator_func(cake):  # decorator function which will recieve the actual function
    def wrapper_fun():     # wrapper function will add extra functionality to the actual function
        print("Buy ingredient for cake") # Extra functionality-1
        cake()
        print("Put the cherry on the top") # Extra functionality-2
    return wrapper_fun

def cake():  # Actual function which you to decorate
    print("Take Wheat Flour")
    print("Milk some sugar and milk")
    print("Add Egg and baking soda")
    print("Put in the oven")
    

updated_func = decorator_func(cake) # passing the actual function to decorator function
updated_func()  # decorated function
```

# The @ symbol for defining decorators

```python
def decorator_func1(dish):  # decorator function which will recieve the actual function
    def wrapper_fun():     # wrapper function will add extra functionality to the actual function
        dish()
        print("Put the cherry on the top")
        print("Chilly also") # Extra functionality-2
    return wrapper_fun

@decorator_func1
def cake():  # Actual function which you to decorate
    print("Take Wheat Flour")
    print("Milk some sugar and milk")
    print("Add Egg and baking soda")
    print("Put in the oven")

cake()
```

# Arithmatic operator overloading

```python
class Dosa:
    def __init__(self, variety, price):
        self.variety = variety
        self.price = price

    def __add__(self,second_obj):
        return self.price + second_obj.price
    def __sub__(self,second_obj):
        return self.price - second_obj.price
    def __mul__(self,second_obj):
        return self.price * second_obj.price
    def __truediv__(self,second_obj):
        return self.price / second_obj.price
    def __floordiv__(self,second_obj):
        return self.price // second_obj.price
    def __mod__(self,second_obj):
        return self.price % second_obj.price
    def __pow__(self,second_obj):
        return self.price ** second_obj.price    
    
        
d1 = Dosa("Masala Dosa", 50)
d2 = Dosa("Allu Dosa", 75)

# print the price of d1 and d2
print(d1 + d2) # add
print(d1 - d2) # sub
print(d1 * d2) # mul
print(d1 / d2) # truediv
print(d1 // d2) # floordiv
print(d1 % d2) # mod
print(d1 ** d2) # pow
```
