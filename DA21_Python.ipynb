# delete operation on set(remove(),discard(),pop(),clear(),del)

```python
lunch_set = {"Two much Rice", "5 Meals","tomato","curd rice","Burger","Potato","Lemon rice"}

# remove() : delete the elements by its  value
lunch_set.remove("5 Meals")
print(lunch_set)

# remove() : function will show the "error" if item don't present in the set.
```

```python
# discard() : delete the elements by its  value
lunch_set.discard("Lemon rice")
print(lunch_set)

# discard() : function will NOT raise any "error" even though the element is not present in the set
```

```python
lunch_set.discard("5 Meals")
print(lunch_set)
```

```python
# pop() : pop will delete the random element
lunch_set.pop()
print(lunch_set)


lunch_set.pop()
print(lunch_set)
```

```python
# clear() : delete all the items

lunch_set.clear()
print(lunch_set)
```

```python
del lunch_set
```
